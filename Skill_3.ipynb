```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/Colab/CGPA_Regression_Dataset.csv')
df.head()
```

```python
X = df[['Study_Hours']].values
y = df['CGPA'].values
```

```python
import numpy as np

m = 0
b = 0

learning_rate = 0.01
epochs = 1000

n = len(X)

for i in range(epochs):
    y_pred = m * X.flatten() + b

    dm = (-2/n) * np.sum(X.flatten() * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db

print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
y_pred = m * X.flatten() + b

print("Actual CGPA:")
print(y)

print("Predicted CGPA:")
print(y_pred)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, label="Regression Line")
plt.xlabel("Study Hours")
plt.ylabel("CGPA")
plt.title("Gradient Descent - CGPA Regression")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("MSE:", mse)
print("R2 Score:", r2)
```
